## Subplots & legends

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
days = [1,2,3,4,5]
```

```python
original_price = [10,20,30,40,50]
```

```python
predicted_price = [8,21,35,38,47]
```

```python
plt.plot(days,original_price, 'b-o', label='Original Price')
plt.plot(days,predicted_price, 'r-o', label ='Predicted Price')
plt.legend()
plt.title('Plot with legends')
plt.xlabel('Weekdays')
plt.ylabel('Price')
```

## Sub Plots:

## First subplot - original price

```python
plt.subplot(3,1,1)
plt.plot(days,original_price, 'b-o', label='Original Price')
plt.title('Original Prices Over time')
plt.xlabel('Weekdays')
plt.ylabel('price')
plt.legend()

## Second subplot - predicted price
plt.subplot(3,1,2)
plt.plot(days,predicted_price, 'r-o', label='Predicted Price')
plt.title('Predicted Prices Over time')
plt.xlabel('Weekdays')
plt.ylabel('price')
plt.legend()

## 3rd subplot - differecnce between original and predicted price
price_difference = [o-p for o,p in zip(original_price, predicted_price)]
plt.subplot(3,1,3)
plt.bar(days,price_difference, color='purple', label='Price Difference')
plt.title('Differecnce between original and predicted price')
plt.xlabel('Days')
plt.ylabel('Price Difference')
plt.legend()


plt.tight_layout()
```

# Saving the figure

```python
plt.plot(days,original_price, 'b-o', label='Original Price')
plt.plot(days,predicted_price, 'r-o', label ='Predicted Price')
plt.legend()
plt.title('Plot with legends')
plt.xlabel('Weekdays')
plt.ylabel('Price')

plt.savefig('example_prediction.png', dpi=300) # save the figure
```
